```python
import matplotlib.colors as mcolors
import matplotlib.transforms as mtransforms

def plot_field_columns(fig, ax, fields, fields_id, iteration,
                       Xmesh, Ymesh, Xgrid, Ygrid,
                       func, field_names, colormap, make_formatter):
    """
    Plot exact, computed, and residual fields (with colorbars) for each field.

    Parameters:
      - fig: matplotlib figure
      - ax: 2D array of axes with shape (3, 1+n_fields)
      - fields: list (or dict) of field data
      - fields_id: list of field ids to plot
      - iteration: current iteration index
      - Xmesh, Ymesh: meshgrid arrays for plotting (edges, n+1 x n+1)
      - Xgrid, Ygrid: meshgrid arrays for evaluation (centers, n x n)
      - func: function returning the exact field (used to compute field differences)
      - field_names: list of field names
      - colormap: colormap to use
      - make_formatter: function to create scalar formatter for colorbars

    Returns:
      (ims_ref, ims_field, ims_res)
    """
    ims_ref, ims_field, ims_res = [], [], []

    # Flatten evaluation grid for model calls
    eval_points = np.hstack((Xgrid.reshape(-1, 1), Ygrid.reshape(-1, 1)))
    fields_exact_all = func(eval_points).reshape(Xgrid.shape + (-1,))

    for i, field_id in enumerate(fields_id):
        # --- Exact solution (top row) ---
        ax_exact = ax[0][i]
        field_exact = fields_exact_all[..., field_id]
        field_norm = mcolors.Normalize(vmin=field_exact.min(), vmax=field_exact.max())
        im_ref = pcolor_plot(ax_exact, Xmesh, Ymesh, field_exact, field_names[field_id] + "*",
                             colormap=colormap, norm=field_norm)
        ims_ref.append(im_ref)

        # Colorbar for exact
        pos = ax_exact.get_position()
        cax_pos = mtransforms.Bbox.from_bounds(pos.x0 + pos.width*0.05, pos.y1 + 0.04,
                                               pos.width*0.9, 0.01)
        cax = fig.add_axes(cax_pos)
        cbfield = fig.colorbar(im_ref, cax=cax, orientation='horizontal', format=make_formatter())
        cbfield.ax.xaxis.set_ticks_position('top')

        # --- Predicted field (middle row) ---
        ax_field = ax[1][i]
        field_pred = np.array(fields[field_id][iteration]).reshape(Xgrid.shape)
        im_field = pcolor_plot(ax_field, Xmesh, Ymesh, field_pred, field_names[field_id],
                               colormap=colormap, norm=field_norm)
        ims_field.append(im_field)

        # --- Residual (bottom row) ---
        ax_res = ax[2][i]
        field_diff = field_pred - field_exact
        diff_norm = mcolors.Normalize(vmin=-np.abs(field_diff).max(), vmax=np.abs(field_diff).max())
        im_res = pcolor_plot(ax_res, Xmesh, Ymesh, field_diff, f"{field_names[field_id]} - {field_names[field_id]}*",
                             colormap='coolwarm', norm=diff_norm)
        ims_res.append(im_res)

        pos = ax_res.get_position()
        cax_pos = mtransforms.Bbox.from_bounds(pos.x0 + pos.width*0.05, pos.y0 - 0.03,
                                               pos.width*0.9, 0.01)
        cax = fig.add_axes(cax_pos)
        fig.colorbar(im_res, cax=cax, orientation='horizontal', format=make_formatter())

    return ims_ref, ims_field, ims_res
```

```python
from matplotlib.patches import Polygon
from matplotlib.colors import to_rgba

def plot_DIC_region(ax, DIC_region, coordMap, n_edge=50, **kwargs):
    """
    Plot the DIC region on a curvilinear mesh, avoiding straight edges.

    Parameters
    ----------
    ax : matplotlib Axes
        Axes on which to plot the region.
    DIC_region : list or array [x_min, x_max, y_min, y_max]
        DIC region coordinates in grid/logical space.
    coordMap : callable
        Function mapping logical coordinates to mesh coordinates.
    n_edge : int
        Number of points to sample per edge for smooth curvilinear borders.
    kwargs : dict
        Additional keyword arguments for Polygon (edgecolor, facecolor, lw, etc.)
    """
    x_min, x_max, y_min, y_max = DIC_region

    # Sample points along each edge
    top_edge = np.column_stack([np.linspace(x_min, x_max, n_edge), np.full(n_edge, y_max)])
    right_edge = np.column_stack([np.full(n_edge, x_max), np.linspace(y_max, y_min, n_edge)])
    bottom_edge = np.column_stack([np.linspace(x_max, x_min, n_edge), np.full(n_edge, y_min)])
    left_edge = np.column_stack([np.full(n_edge, x_min), np.linspace(y_min, y_max, n_edge)])

    # Concatenate edges in order
    rect_points = np.vstack([top_edge, right_edge, bottom_edge, left_edge])

    # Map points using coordMap
    rect_points_mapped = np.array(jax.vmap(coordMap)(rect_points))

    # Create polygon
    poly = Polygon(rect_points_mapped, closed=True, **kwargs)
    ax.add_patch(poly)
    return poly


def plot_slice(ax, field_id, field_names, fields, Xgrid, Ygrid, Xmesh, Ymesh,
               coordMap, L_max, slice_value, slice_min=0.0, slice_max=None, direction='vertical'):

    # Field
    field_name = field_names[field_id]
    field = fields[field_id][iteration].reshape(Xgrid.shape)
    field_exact = func(np.stack((Xgrid.ravel(), Ygrid.ravel()), axis=1))[:, field_id].reshape(Xgrid.shape)

    # Interpolators
    x_center = Xgrid[:,0]
    y_center = Ygrid[0,:]
    interp_field = RegularGridInterpolator((x_center, y_center), field, bounds_error=False, fill_value=np.nan)
    interp_field_exact = RegularGridInterpolator((x_center, y_center), field_exact, bounds_error=False, fill_value=np.nan)

    # Slice on edge mesh
    if direction == 'vertical':
        # pick nearest X column
        idx = np.argmin(np.abs(Xmesh[:,0] - slice_value))
        Xslice = Xmesh[idx,:]
        Yslice = Ymesh[idx,:]
        mask = (Yslice >= slice_min) & (Yslice <= (slice_max or L_max))
    elif direction == 'horizontal':
        # pick nearest Y row
        idx = np.argmin(np.abs(Ymesh[0,:] - slice_value))
        Xslice = Xmesh[:,idx]
        Yslice = Ymesh[:,idx]
        mask = (Xslice >= slice_min) & (Xslice <= (slice_max or L_max))
    else:
        raise ValueError("direction must be 'vertical' or 'horizontal'")

    Xslice = Xslice[mask]
    Yslice = Yslice[mask]

    # Sort for plotting along the slice
    if direction == 'vertical':
        sort_idx = np.argsort(Yslice)
    else:
        sort_idx = np.argsort(Xslice)
    Xslice, Yslice = Xslice[sort_idx], Yslice[sort_idx]

    # Map cell centers along the slice
    if direction == 'vertical':
        idx_grid = np.argmin(np.abs(Xgrid[:,0] - slice_value))
        Grid_slice = np.column_stack((Xgrid[idx_grid,:], Ygrid[idx_grid,:]))
    else:
        idx_grid = np.argmin(np.abs(Ygrid[0,:] - slice_value))
        Grid_slice = np.column_stack((Xgrid[:,idx_grid], Ygrid[:,idx_grid]))

    # Map to physical mesh coordinates
    Mesh_slice = jax.vmap(coordMap)(Grid_slice)
    mask_grid = ( (Mesh_slice[:,1] >= slice_min) & (Mesh_slice[:,1] <= (slice_max or L_max)) ) \
                if direction == 'vertical' else \
                ( (Mesh_slice[:,0] >= slice_min) & (Mesh_slice[:,0] <= (slice_max or L_max)) )

    grid_points = Grid_slice[mask_grid]
    mesh_points = Mesh_slice[mask_grid]

    # Interpolate field values
    field_slice_vals = interp_field(grid_points)
    field_exact_slice_vals = interp_field_exact(grid_points)

    # --- Plot ---
    # 2D field
    im1 = pcolor_plot(ax[0], Xmesh, Ymesh, field_exact, field_name + "*", colormap=cmap)
    fig.colorbar(im1, ax=ax[0], orientation='vertical', format=make_formatter(), location='left', shrink=0.92, pad=0.05)

    DIC_region = run_config["data"]["DIC_region"]
    # Overlay DIC region
    plot_DIC_region(ax[0], DIC_region, coordMap, edgecolor='w', facecolor=to_rgba('w', alpha=0.1), lw=0.8, label='DIC region')
    # ax[0].legend(loc='upper right')
    x_pos = 50    # adjust to where you want the text
    y_pos = 82
    ax[0].text(x_pos, y_pos, 'DIC region', color='w', fontsize=15, fontweight='bold', ha='center', va='bottom')
    # Overlay curvilinear slice
    ax[0].plot(Xslice, Yslice, 'r-', lw=1.5)

    # 1D slice plot
    if direction == 'vertical':
        ax[1].plot(mesh_points[:,1], field_slice_vals, 'b-', label='PINN')
        ax[1].plot(mesh_points[:,1], field_exact_slice_vals, 'r--', label='FEM')
        ax[1].set_xlabel('Y (mm)')
    else:
        ax[1].plot(mesh_points[:,0], field_slice_vals, 'b-', label='PINN')
        ax[1].plot(mesh_points[:,0], field_exact_slice_vals, 'r--', label='FEM')
        ax[1].set_xlabel('X (mm)')
    ax[1].set_ylabel(field_name + ' (MPa)')
    ax[1].legend()

fig, ax = plt.subplots(3, 4, figsize=(15, 10), dpi=150)

for i, field_id in enumerate([5,6,7]):

    plot_slice(ax[i][:2], field_id=field_id, field_names=field_names_tex, fields=fields,
               Xgrid=Xgrid, Ygrid=Ygrid, Xmesh=Xmesh, Ymesh=Ymesh,
               coordMap=coordMap, L_max=L_max,
               slice_value=50.0, slice_min=0*L_max, slice_max=1*L_max,
               direction='horizontal')
    
    plot_slice(ax[i][2:],field_id=field_id, field_names=field_names_tex, fields=fields,
               Xgrid=Xgrid, Ygrid=Ygrid, Xmesh=Xmesh, Ymesh=Ymesh,
               coordMap=coordMap, L_max=L_max,
               slice_value=100.0, slice_min=0.2*L_max, slice_max=0.8*L_max,
               direction='vertical')

plt.tight_layout(pad=0.5, w_pad=0.5, h_pad=0.5)
# plt.show()
print(f"Fig width: {fig.get_figwidth()}, set it to {fig.get_figwidth()*printed_font_size/title_font_size:.2f} < 6.34 (A4 with margins) in latex to have a printed font size of {printed_font_size} for titles and {axes_font_size*printed_font_size/title_font_size:.2f} for the axis")


```

```python
from builtins import enumerate


def animate_frame(i, ims_field, ims_res, ims_weights,
                  lines, scatters, steps, time_unit, step_type, metrics, fields, fields_weights, cb_weights, func, Xgrid, Ygrid, field_names, fields_id):

    plt.suptitle(f"{step_type}: {steps[i]:.0f}{f' {time_unit}' if step_type=='time' else ''}", fontsize=16)
    
    # Update metric plots
    lines, scatters = update_metrics(i, lines, scatters, steps, metrics)

    # Update weight plots
    if len(ims_weights) > 0:
        for j, im in enumerate(ims_weights):
            im.set_array(fields_weights[j][i, :].reshape(Xgrid.shape).ravel())
            norm =  mcolors.Normalize(vmin=fields_weights[j][i, :].min(), vmax=fields_weights[j][i, :].max())
            im.set_norm(norm)
            cb_weights[j].ax.yaxis.set_ticks_position('left')

    # Update field and residual plots for each field
    for idx, fid in enumerate(fields_id):
        field = np.array(fields[fid][i]).reshape(Xgrid.shape)
        field_exact = func(np.hstack((Xgrid.reshape(-1, 1), Ygrid.reshape(-1, 1))))[:, fid].reshape(Xgrid.shape)
        field_diff = field - field_exact
        
        ims_field[idx].set_array(field.ravel())
        ims_res[idx].set_array(field_diff.ravel())
        norm =  mcolors.Normalize(vmin=field_diff.min(), vmax=field_diff.max())
        ims_res[idx].set_norm(norm)
    
    return ims_field, ims_res, ims_weights, lines, scatters
```
